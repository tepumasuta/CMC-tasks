```python
import sympy as smp
from sympy import *
```

```python
m0, m1, m2, a0, a1, a2, g, t1, t2 = smp.symbols('m0 m1 m2 a0 a1 a2 g t1 t2', real=True)
```

```python
smp.solve([m0*a0 - t1, -m1*g+t2-m1*a1, -m2*g+t2-m2*a2, 2*a0+a1+a2, t1-2*t2], [a0, a1, a2, t1, t2])
```

```python
(-g*m0*m1 + g*m0*m2 - 4*g*m1*m2)/(m0*m1 + m0*m2 + 4*m1*m2)
```

```python
t1, t2, t3, m, M, a, A, f = smp.symbols('t1, t2, t3, m, M, a, A, f', real=True)
```

```python
smp.solve([t1-2*t2, t3-2*t2, -M*g+t3-M*A, -m*g+t2-m*a, 2*A+a, f-t1-t2], [t1, t2, t3, a, A, f])
```

```python
m1, m2, v1, v2, vn1, vn2 = smp.symbols('m1 m2 v1 v2 vn1 vn2', real=True)
```

```python
smp.solve([m1*v1-m2*v2-m1*vn1+m2*vn2, m1*v1**2/2+m2*v2**2/2-m1*vn1**2/2-m2*vn2**2/2], [vn1, vn2])
```

```python
((m1*v1 - m2*v2 + m2*(-2*m1*v1 - m1*v2 + m2*v2)/(m1 + m2))/m1).simplify()
```

```python
((-2*m1*v1 - m1*v2 + m2*v2)/(m1 + m2)).simplify()
```

```python
m, M, v0, v1, v2 = smp.symbols('m M v0 v1 v2', real=True)
```

```python
smp.solve([M*v2+m*v1-m*v0, M*v2**2/2+m*v1**2/2-m*v0**2/2], [v1, v2])
```

```python
((-2*M*m*v0/(M + m) + m*v0)/m).simplify()
```

```python
(2*m*v0/(M + m)).simplify()
```

```python
m, M, v0, v1, v2, mu, g, t, l, vn1, vn2, vx = smp.symbols('m M v0 v1 v2 mu g t l vn1 vn2 vx', real=True)
```

```python
smp.solve([m*v0-(m+M)*vx, m*v0**2/2-(m+M)*vx**2/2+2*mu*m*g*l], [v0, vx])
```

```python
m, u, v, vn, a = smp.symbols('m u v vn a', real=True)
```

```python
smp.solve([m*v-3*m*u*smp.sqrt(2)/2-m*vn*smp.cos(a),
           m*v**2-3*m*u**2-m*vn**2,
           3*m*u*smp.sqrt(2)/2-m*vn*smp.sin(a)], [u, vn, a])
```

```python
smp.asin(3*sqrt(10)*smp.Abs(v)/(10*v)).simplify()
```

```python
M, m, u, v, c = smp.symbols('M m u v c', real=True)
```

```python
smp.solve([M*u/smp.sqrt(1-u**2/c**2)-m*v/smp.sqrt(1-v**2/c**2),
           M*c**2/smp.sqrt(1-u**2/c**2)-m*c**2/smp.sqrt(1-v**2/c**2)-m*c**2], [M, u])
```

```python
(c**2*v*(sqrt((c**2 - v**2)/c**2) + 1)/(2*c**2*sqrt((c**2 - v**2)/c**2) + 2*c**2 - v**2)).simplify()
```
