```python
import sympy as smp
from sympy import *
```

```python
u0, v, u = smp.symbols('u_0 v u', real=True)
v_v = smp.Matrix([-v, 0])
u_v = smp.Matrix([1, 1]) / smp.sqrt(2) * u
eq1 = (u_v - v_v).norm() - u0
smp.solve([eq1], u)
```

```python
v1, n1, v2, n2, n0, vd = smp.symbols('v_1 n_1 v_2 n_2 n_0 vd', real=True)
v_v1 = smp.Matrix([v1, 0]) / 18 * 5
v_v2 = smp.Matrix([v2, 0]) / 18 * 5
vd_v = smp.Matrix([0, -vd])
vd_vp1 = vd_v - v_v1
vd_vp2 = vd_v - v_v2
smp.solve([(vd_vp1).norm() / (vd_vp2).norm() - n1 / n2], vd)
```
